```python
import numpy as np
from utils import sir_model, sir_solver
import matplotlib.pyplot as plt
import os

# Constants
y0 = [1995, 5, 0]
days = 160
dt = 0.1
beta = 0.125*1e-3
gamma = 0.1
N = 2000
```

```python
t, y = sir_solver(sir_model, y0, days, dt, method="runge_kutta", beta=beta, gamma=gamma)

S = y[:, 0]
I = y[:, 1]
R = y[:, 2]
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(t, S, label="Susceptible")
ax.plot(t, I, label="Infected")
ax.plot(t, R, label="Recovered")
ax.set_xlabel("Time (days)")
ax.set_ylabel("Population")
ax.set_title("SIR Model (N=2000)")
ax.legend()
ax.grid()

os.makedirs("figures", exist_ok=True)
fig.savefig("figures/sir_model.png")

plt.show()
```

```python
# Find the peak of the infected population
peak_infected = np.max(I)
peak_day = t[np.argmax(I)]
print(f"Peak Infected: {peak_infected} on Day: {peak_day}")
```

```python
# Mark the peak in the plot
ax.axvline(peak_day, color="red", linestyle="--", linewidth=1,
           label=f"Peak: day {peak_day:.1f} ({peak_infected:.0f} infected)")
ax.legend()
fig.savefig("figures/sir_model_peak.png")
fig
```
