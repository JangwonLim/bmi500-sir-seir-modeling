```python
import numpy as np
from utils import sir_model, sir_solver

# Constants
y0 = [1995, 5, 0]
days = 160
dt = 0.1
beta = 0.125*1e-3
gamma = 0.1
N = 2000
```

```python
# copied from partB
t, y = sir_solver(sir_model, y0, days, dt, method="runge_kutta", beta=beta, gamma=gamma)

S = y[:, 0]
I = y[:, 1]
R = y[:, 2]

# Find the peak of the infected population
peak_infected = np.max(I)
peak_day = t[np.argmax(I)]
print(f"Peak Infected: {peak_infected:.0f} on Day: {peak_day}")
```

```python
# basic reproduction number
R0 = beta * N / gamma
print(f"Basic Reproduction Number (R0): {R0}")
```

```python
# The number of people who are never infected by the end of the simulation
never_infected = S[-1]
print(f"Number of people never infected: {never_infected:.0f}")
```
